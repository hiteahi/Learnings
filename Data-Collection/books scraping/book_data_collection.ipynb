# Data Collection using requests, BeautifulSoup

## Notebook Summary

This notebook demonstrates web scraping for data collection using Python libraries `requests` and `BeautifulSoup`. It scrapes book information (title, price, and rating) from the Books to Scrape website (https://books.toscrape.com).

### Key Steps:
1. **Setup**: Import necessary libraries (`requests`, `BeautifulSoup`, `pandas`).
2. **Status Check**: Verify website accessibility by checking the HTTP response status code for the first page.
3. **Download HTML Pages**: Fetch and save HTML content from 50 catalog pages locally in the `htmls/` directory.
4. **Parse HTML Content**: Extract and print book titles from the first downloaded page as an initial parsing example.
5. **Collect Data from One Page**: Parse detailed book information (title, price, rating) from the first page and create a pandas DataFrame.
6. **Collect Data from All Pages**: Scale the data extraction to all 50 pages, aggregating information for all books into a comprehensive DataFrame.
7. **Save Data**: Export the collected data to a CSV file (`books.csv`) for further analysis or use.

### Output:
- A pandas DataFrame displaying book titles, prices (in GBP), and star ratings.
- A CSV file containing the scraped data with columns: Title, Price, Rating.

This notebook serves as a complete example of web scraping, HTML parsing, data extraction, and basic data processing using Python.

## Setup: Importing Libraries

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd, numpy as np
```

## Step 1: Checking Website Response

**Common Status Codes**
- 200: OK (Success)
- 404: Not Found
- 403: Forbidden
- 500: Internal Server Error

```python
url = "https://books.toscrape.com/catalogue/page-1.html"

response = requests.get(url)

print(response.status_code)
```

## Step 2: Downloading HTML Pages

```python
for num in range(1, 51):
    url = f"https://books.toscrape.com/catalogue/page-{num}.html"
    response = requests.get(url)
    with open(f"htmls/page-{num}.html", "w", encoding="utf-8") as file:
        file.write(response.text)
        print(f"Saved page-{num}.html")
```

## Step 3: Parsing HTML Content

```python
with open("htmls/page-1.html") as file:
    content = file.read()
# parse the html content using BeautifulSoup
soup = BeautifulSoup(content, "html.parser")

# find all h3 tags in the html
h3s = soup.find_all("h3")

for h3 in h3s:
    # get the title attribute of the anchor tag inside the h3
    print(h3.a["title"])
```

## Step 4: Collecting Data from One Page

```python
articles = soup.find_all("article", class_="product_pod")
items = []
for article in articles:
    # get the title attribute of the anchor tag inside the h3
    title = article.h3.a["title"] 
    # get the text of the p tag with class price_color and split it to get the price
    price = article.select_one("p.price_color").text.split("£")[1] 
    # get the second class of the article tag which is the rating
    rating = article.p["class"][1]

    # append the title, price and rating to the items list
    items.append([title, price, rating])

# create a dataframe from the items list and set the column names to Title, Price and Rating
df = pd.DataFrame(items, columns=["Title", "Price", "Rating"])
df
```

## Step 5: Collecting Data from All Pages

```python
items = []
for num in range(1, 51):
    with open(f"htmls/page-{num}.html") as file:
        content = file.read()
        soup = BeautifulSoup(content, "html.parser")

        # Each book is inside an 'article' tag with class 'product_pod'
    books = soup.find_all('article', class_='product_pod')
    
    for book in books:
        # 1. Book Name (stored in the 'title' attribute of the <a> tag inside <h3>)
        name = book.h3.a['title']
        
        # 2. Price (stored in the text of a <p> tag with class 'price_color')
        price = book.find('p', class_='price_color').text.split("£")[1]  # Remove the '£' symbol and get the price
        
        # 3. Rating (stored as a class name like 'star-rating Three')
        # We take the second class name which is the actual rating word
        rating_classes = book.find('p', class_='star-rating')['class']
        rating = rating_classes[1] 

        items.append([name, price, rating])

# create a dataframe from the items list and set the column names to Title, Price and Rating
df = pd.DataFrame(items, columns=["Title", "Price", "Rating"])
df
```

## Step 6: Converting Ratings to Numbers

```python
rating_mapping = {
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
}
df["Rating"] = df["Rating"].replace(rating_mapping)
```

## Step 7: Saving Data to CSV

```python
df.to_csv("books_data.csv", index=False)
```

## Conclusion

This notebook successfully demonstrates a complete web scraping workflow:
- Fetched HTML content from 50 pages of the Books to Scrape website.
- Extracted book details (title, price, rating) for approximately 1000 books.
- Processed the data by converting ratings to numeric values.
- Saved the cleaned data to a CSV file for further analysis.

The resulting `books.csv` file contains structured data ready for data analysis, visualization, or machine learning tasks.
