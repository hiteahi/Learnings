## Pivot Tables in Pandas

Pivot tables summarize data by transforming rows into structured
analytical tables using aggregation functions.

Pivot tables internally use groupby and aggregation but reshape results into a matrix format.

## How it works:
`pivot()` takes a long-format DataFrame and turns it into a wide-format
DataFrame by specifying which columns will become the new columns, the
rows, and the values.

```python
import pandas as pd, numpy as np
```

```python
df = pd.read_csv("data.csv")
df.head(), df.tail()
```

## Basic Pivot Table

```python
df.pivot_table(
    values='salary',
    index='department',
    aggfunc='mean'
              )
```

## Pivot with Rows and Column

```python
df.pivot_table(
    values='salary',
    index='department',
    columns='city',
    aggfunc='mean'
              )
```

This type of table is commonly used in financial and HR reports
to compare performance across departments and locations.

## Multiple Aggregations
Multiple aggregation functions create hierarchical column labels.

```python
df.pivot_table(
    values='salary',
    index='department',
    aggfunc=['mean', 'max', 'min', 'count']
              )
```

pivot_table is built on groupby internally,
making it efficient for large datasets.

## Multiple Value Columns

```python
df.pivot_table(
    values=['salary', 'age'],
    index='department',
    aggfunc='mean'
              )
```

## Handling Missing Values in Pivot

```python
df.pivot_table(
    values='salary',
    index='department',
    columns='city',
    aggfunc='mean',
    fill_value=0
              )
```

## Pivot vs GroupBy
| GroupBy                  | Pivot Table                |
| ------------------------ | -------------------------- |
| Flexible                 | Structured output          |
| Good for analysis        | Good for reporting         |
| Returns Series/DataFrame | Returns matrix-style table |


## Summary
- pivot_table combines grouping and reshaping
- It supports aggregation
- It handles duplicate entries
- It is ideal for summary reports
