# GroupBy and Aggregations

GroupBy operations allow us to summarize and analyze data
by categories such as department, city, or time.

This analysis focuses on understanding patterns in the data
using aggregation techniques such as averages and counts.

```python
import pandas as pd, numpy as np
```

```python
df = pd.read_csv("data.csv")
df.head(), df.tail()
```

## Basic GroupBy
Group by Department, then calculate average Salary for each group.

```python
df.groupby('department')['salary'].mean()
```

## Common Aggregation Functions

### Average Salary per City

```python
df.groupby("city")["salary"].mean() 
```

### Total Salary by Department

```python
df.groupby("department")["salary"].sum() 
```

### Total Employees by City

```python
df.groupby("city")["salary"].count()
```

### Maximum Salary per City

```python
df.groupby("city")["salary"].max()
```

### Minimum Salary per City

```python
df.groupby("city")["salary"].min()
```

## Multiple Aggregations

```python
df.groupby('department')['salary'].agg(['mean', 'min', 'max', 'count'])
```

```python
df.groupby('department')['salary'].agg(['mean', 'median'])
```

## GroupBy on Multiple Columns

```python
df.groupby(['department', 'city'])['salary'].mean()
```

## Custom Aggregations with .agg()

```python
df.groupby('department').agg({
    'salary': 'mean',
    'age': 'median'
})
```

## Named Aggregation (Cleanest way)

```python
# Better than dictionary agg
df.groupby('department').agg(
    avg_salary=('salary', 'mean'),
    max_age=('age', 'max'),
    employee_count=('id', 'count')
                            )
```

## Time-based GroupBy

```python
df['year'] = pd.to_datetime(df['joining_date']).dt.year
df.groupby('year')['salary'].mean()
```

## Reset Index

```python
df.groupby('department')['salary'].mean().reset_index()
```

## Summary
- .groupby() helps you summarize large datasets by category
- Aggregations simplify complex datasets
- Use mean() , sum() , count() , .agg() for custom metrics
