# Sorting & Ranking
Sorting and ranking allow us to organize data and identify
top or bottom performers in a dataset.

```python
import pandas as pd, numpy as np
```

```python
df = pd.read_csv("Data.csv")
df.head(15)
```

## Sorting

### Sort by Values

```python
df.sort_values("name").head(10) # Ascending sort
```

```python
df.sort_values("name", ascending=False).head(10) # Descending
```

```python
df.sort_values(["department", "age"], inplace=True) # Sort by multiple columns
df.head(10)
```

`df.sort_values(["department", "age"])` sorts the DataFrame first by the **“department”** column, and
if there are ties (i.e., two or more rows with the same “department”), it will sort by the
**“age”** column.


## Sort by Index

The `df.sort_index()` function is used to sort the DataFrame based on its index
values. If the index is not in a sequential order (e.g., you have dropped rows or
performed other operations that change the index), you can use sort_index() to
restore it to a sorted order

```python
df.sort_index().head(15)
```

## Ranking
Ranking allows organizations to identify top performers,
high-value customers, or priority segments.

When ties occur, the ranking method determines whether
gaps appear in the ranking sequence.

### Average Method(Default)
Assigns the average rank of the tied values. Ranks may have decimal places.

```python
df["Rank"] = df["salary"].rank(ascending=False) # 9.5 (average of 9 and 10)
df.sort_values("Rank").head(10)
```

### First Method
Assigns unique ranks based on the order of appearance in the array. No ranks are skipped.

```python
df["Rank"] = df["salary"].rank(method="first", ascending=False) # 1, 2, 3 (assigned in order of appearance)
df.sort_values("Rank").head(10)
```

### Dense Method
Assigns ranks using consecutive integers with no gaps, even in the presence of ties.

```python
df["Rank"] = df["salary"].rank(method="dense", ascending=False) # 9, 9 (next rank is 10)
df.sort_values("Rank").head(10)
```

### Max Method
Assigns the highest rank in the group to all tied values. Ranks after the tie are skipped.

```python
df["Rank"] = df["salary"].rank(method="max", ascending=False) # Skip 9 (next rank is 10)
df.sort_values("Rank").head(10)
```

### Min Method
Assigns the lowest rank in the group to all tied values. Ranks after the tie are skipped. (e.g., 2 for a tie in positions 2 and 3).

```python
df["Rank"] = df["salary"].rank(method="min", ascending=False) # Skip 10 (next rank is 11)
df.sort_values("Rank").head(12)
```

## Ranking Within Groups

```python
df['dept_rank'] = df.groupby('department')['salary'] \
                     .rank(ascending=False)
df.sort_values('dept_rank').head(10)
```

## Top N Selection
nlargest() is more efficient than sorting the entire dataset
when only top records are required.


### Top 3 by salary

```python
df.nlargest(3, 'salary')
```

### Last Top 3 by salary

```python
df.nsmallest(3, 'salary')
```

## Summary
- sort_values organizes data
- rank assigns order positions
- groupby + rank enables category-based ranking
- nlargest is efficient for top-N queries
